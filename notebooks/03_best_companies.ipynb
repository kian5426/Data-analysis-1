# 03 – Best Companies
For non-dark chocolates (cocoa ≤ 70%), find the top-rated companies and the bars they produce.

## Load priced data

```python
import numpy as np
import pandas as pd

df = pd.read_csv('../data/processed/chocolate_price.csv')
df.head()
```

## Select non-dark chocolates

```python
df = df.rename(columns={'Cocoa Percent':'Percent'})
df = df[df.Percent <= 70].reset_index()
df = df.rename(columns={'Percent':'Cocoa Percent'})
df.drop(['index'], axis=1, inplace=True)
df.head()
```

## Mean rating per company and year
Pivot to a company × review-year table.

```python
companies = df.groupby(['Company','Review Date'])['Rating'].mean().unstack()
companies.head()
```

```python
companies.columns.name = None
companies.index.name = None
```

## Mean rating since 2012
Average each company's yearly ratings from 2012 onward.

```python
mean_ratings = companies.loc[:, 2012:].mean(axis=1)
mean_ratings.head()
```

## Top 10 companies
Rank by mean rating; ties are broken alphabetically (stable sort on the sorted index).

```python
best_ratings = mean_ratings.sort_values(ascending=False, kind='mergesort')
best_ratings = best_ratings.sort_index(kind='mergesort')
best_ratings = best_ratings.sort_values(ascending=False, kind='mergesort').head(10)
best_ratings
```

## Chocolates to sell
Select all bars made by the top-10 companies.

```python
chocolates_to_sell = df[df['Company'].isin(best_ratings.index)]
chocolates_to_sell = chocolates_to_sell.reset_index(drop=True)
chocolates_to_sell.head()
```

## Total price

```python
priceSum = chocolates_to_sell['price(100g)'].sum()
```

## Save

```python
df.to_csv('../data/processed/normal_chocolates.csv', index = True)
companies.to_csv('../data/processed/companies.csv', index = True)
chocolates_to_sell.to_csv('../data/processed/chocolates_to_sell.csv', index = True)
mean_ratings.to_csv('../data/processed/mean_ratings.csv', index = True)
best_ratings.to_csv('../data/processed/best_ratings.csv', index = True)
```
