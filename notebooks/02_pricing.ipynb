# 02 – Pricing
Estimate a price per 100 g for each bar from its rating and cocoa content, then isolate dark chocolates.

## Load preprocessed data

```python
import numpy as np
import pandas as pd

df = pd.read_csv('../data/processed/chocolate_preprocessed.csv')
df.head()
```

## Rescale ratings
Scale `Rating` so the maximum becomes 100.

```python
a = df['Rating'].max()
df['Rating'] = df['Rating']*100/a
df.head()
```

## Price per 100 g
`price(100g) = Cocoa Percent × Rating × 25`, so a 100% cocoa, top-rated bar costs 250,000.

```python
df['price(100g)'] = df['Cocoa Percent'] * df['Rating'] * 25
df.head()
```

## Select dark chocolates
Dark chocolate is defined as cocoa content > 70%.

```python
dfc = df.copy()
dfc = dfc.rename(columns={'Cocoa Percent':'Percent'})
dark_chocolates = dfc[dfc.Percent > 70].reset_index()
dark_chocolates = dark_chocolates.rename(columns={'Percent':'Cocoa Percent'})
dark_chocolates.drop(['index'], axis=1, inplace=True)
dark_chocolates.head()
```

## Trinitario price adjustment
Flag bars whose `Bean Type` contains *Trinitario* and raise their price by 10%.

```python
has_Trinitario = dark_chocolates['Bean Type'].str.contains('Trinitario', na=False)
```

```python
dark_chocolates.loc[has_Trinitario, 'price(100g)'] *= 1.1
dark_chocolates.head()
```

## Total price of dark chocolates

```python
priceSum = dark_chocolates['price(100g)'].sum()
```

## Save

```python
df.to_csv('../data/processed/chocolate_price.csv', index = False)
dark_chocolates.to_csv('../data/processed/dark_chocolates.csv', index = True)
```
